# Evaluate and display low confidence scores

Run the following command

`python medical_pipeline.py --audio ../data/test_audio/test_audio.mp3 --transcription-only`


```python
import json
import pandas as pd
# Your data object
data = pd.read_json('../output.json')
```

```python

def flag_low_confidence(json_data, threshold=80.0):
    print(f"--- Flagging sentences with confidence below {threshold}% ---")
    low_confidence_items = []
    
    # Iterate through the sentence_confidences list
    for entry in json_data.get("sentence_confidences", []):
        if entry["confidence_percent"] < threshold:
            low_confidence_items.append(entry)
            
            # Print a summary of the flagged item
            print(f"Index: {entry['sentence_index']} | "
                  f"Score: {entry['confidence_percent']}% | "
                  f"Text: \"{entry['text']}\"")
    
    if not low_confidence_items:
        print("No low confidence scores found.")
        
    return low_confidence_items

# Execute the function
flagged_results = flag_low_confidence(data, 60)
```
